```python
from FFNNnormal import FFNN
from AutoDiff import Value
from sklearn.datasets import make_moons, make_blobs
import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = make_moons(n_samples=100, noise=0.1, random_state=10)

y = y*2 - 1 # make y be -1 or 1
# visualize in 2D
plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], c=y, s=20, cmap='jet')
```

```python
model = FFNN(layers=[2, 10, 1], activation_functions=["relu", "linear"], loss_function="mse", batch_size=1, learning_rate=0.1, epoch=1000, verbose=1, random_state=10)
model.weight_initializer(weight_initializer="normal", mean=0, variance=1, seed=10)
model.fit(X, y)
```

```python
h = 0.25
x_min, x_max = X[:, 0].min() - 1, X[:, 0].max() + 1
y_min, y_max = X[:, 1].min() - 1, X[:, 1].max() + 1
xx, yy = np.meshgrid(np.arange(x_min, x_max, h),
                     np.arange(y_min, y_max, h))
Xmesh = np.c_[xx.ravel(), yy.ravel()]
Xmesh = np.c_[xx.ravel(), yy.ravel()]
Z = np.array(model.predict([list(xrow) for xrow in Xmesh]))
Z = Z.reshape(xx.shape)

fig = plt.figure()
plt.contourf(xx, yy, Z, cmap=plt.cm.Spectral, alpha=0.8)
plt.scatter(X[:, 0], X[:, 1], c=y, s=40, cmap=plt.cm.Spectral)
plt.xlim(xx.min(), xx.max())
plt.ylim(yy.min(), yy.max())
```
