```python
from FFNN import FFNN
from AutoDiff import Value
from sklearn.datasets import make_moons, make_blobs
import numpy as np
import matplotlib.pyplot as plt
```

```python
X, y = make_moons(n_samples=1000, noise=0.1, random_state=10)

y = y*2 - 1 # make y be -1 or 1
# visualize in 2D
plt.figure(figsize=(5,5))
plt.scatter(X[:,0], X[:,1], c=y, s=20, cmap='jet')
```

```python
model = FFNN(layers=[2, 16, 1], activation_functions=["linear", "sigmoid", "relu"], loss_function="mse", batch_size=1, learning_rate=0.01, epoch=150, verbose=1, random_state=10)
model.weight_initializer(weight_initializer="normal", mean=1, variance=0, seed=10)
# model.visualize_graph()
```

```python
model.fit(X, y)
```

```python
# Predict the labels using the model
y_pred = model.predict(X)
print(len(X))
print(y_pred)

# Calculate accuracy
accuracy = np.mean(y_pred == y)
print(f"Accuracy: {accuracy * 100:.2f}%")
```
