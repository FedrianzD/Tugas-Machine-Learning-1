```python
import os
from FFNN import FFNN
```

```python
file_names = [os.path.splitext(file)[0] for file in os.listdir('./models') if file.endswith('.pkl')]
file_names.remove('model_sklearn')
file_names.sort()
```

```python
for file_name in file_names:
    model = FFNN.load("models/" + file_name + ".pkl")
    model.visualize_weight_distributions(list(range(len(model.layers))), file_name)
    model.visualize_weight_gradient_distributions(list(range(len(model.layers))), file_name)
    model.visualize_validation_loss(file_name)
    model.visualize_train_loss(file_name)
```
